```python
import polars as pl
import sys
sys.path.append('src')
from data.transform import train_val_test_split_scaffold

# Wczytanie zbioru danych
df = pl.read_parquet('src/datasets/clean.parquet')

total_samples = len(df)
tp_count = df['is_active'].sum()
tn_count = total_samples - tp_count

print("=== CAŁY ZBIÓR ===")
print(f"Całkowita liczba par: {total_samples:,}")
print(f"Aktywne (TP): {tp_count:,} ({tp_count/total_samples*100:.2f}%)")
print(f"Nieaktywne (TN): {tn_count:,} ({tn_count/total_samples*100:.2f}%")

print("=== PODZIAŁ SCAFFOLD SPLIT (0.7, 0.1, 0.2) ===")
train_df, val_df, test_df = train_val_test_split_scaffold(df, proportions=[0.7, 0.1, 0.2])

for name, split_df in zip(["Treningowy", "Walidacyjny", "Testowy"], [train_df, val_df, test_df]):
    s_total = len(split_df)
    s_tp = split_df['is_active'].sum()
    s_tn = s_total - s_tp
    print(f"[{name}]")
    print(f"  Liczba par: {s_total:,} ({s_total/total_samples*100:.1f}% całości)")
    print(f"  Aktywne (TP): {s_tp:,} ({s_tp/s_total*100:.1f}%)")
    print(f"  Nieaktywne (TN): {s_tn:,} ({s_tn/s_total*100:.1f}%)")
```
